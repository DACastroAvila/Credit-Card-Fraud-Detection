First we will just import the necessary libraries

```python
import pandas as pd
import numpy as np
```

In here we will import the data set that will be used

```python
data = pd.read_csv(r"D:\Castro\Documentos\Git\Data\raw\creditcard.csv")
```

Next, we will divide our data set into two new subsets that will be used for training and testing the model. Here we will use 33% of the data for the testing part.

```python
from sklearn.model_selection import train_test_split
x = data.copy()
y = data["Class"]
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.33)
```

This next part of the the code will do the part of implementing the model.

```python
from sklearn.tree import DecisionTreeClassifier
clf = DecisionTreeClassifier()
clf = clf.fit(x_train, y_train)
clf.get_params()
predictions = clf.predict(x_test)
clf.predict_proba(x_test)
```

Finally all of the next prompts are used to measure the accuracy, precisión and other metrics that can be determined with the sklearn library

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,predictions)
```

```python
from sklearn.metrics import confusion_matrix
confusion_matrix(y_test, predictions, labels=[0,1])
```

```python
from sklearn.metrics import precision_score
precision_score(y_test, predictions)
```

```python
from sklearn.metrics import recall_score
recall_score(y_test, predictions)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, predictions, target_names=['Not fraudulent', 'fraudulent']))
```
